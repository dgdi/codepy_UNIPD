## Pandas

- Central tool for data cleaning and analysis in Python
- Often used alongside NumPy, SciPy, statsmodels, scikit-learn, and seaborn/matplotlib
- Adopts NumPy's array-based computing style
    - Focuses on data processing without the need for iteration
- Pandas vs. NumPy: Key Differences
    - Pandas is designed for tabular or heterogeneous data
    - NumPy is best for homogeneously typed numerical array data
- Focuses on two main data structures: Series and DataFrame.
    - Series is a one-dimensional array-like object
    - DataFrame is a two-dimensional array-like object

## Pandas Basics

### Series in Pandas

- One-dimensional array-like object
    - Contains a sequence of values and an associated array of data labels called `index`
- Values must be of the same type
- Index labels can be a (possibly non-sequential) integer sequence or a string sequence
    - Index allows to reference values in the Series
    - One of the most important features of Pandas
        - Makes it easy to perform operations based on index labels
        - E.g., group operations, align data of different objects for computation or merging

### Series in Pandas


````python
import numpy as np
import pandas as pd
ser1 = pd.Series([4, 7, -5, 3]) 
ser2 = pd.Series([4, 7, -5, 3], index=['d', 'b', 'a', 'c'])
print(f"ser1 = pd.Series([4, 7, -5, 3]) creates a Series with default index: \n{ser1}\n")
print(f"ser2 = pd.Series([4, 7, -5, 3], index=['d', 'b', 'a', 'c']) creates a Series with custom index: \n{ser2}")
````

### Index Updating
- The index of an existing Series can be modified by assignment
    - `series.index = new_index`
    - New index must have the same length as the old one
        - Usually a list of ordered values or a transformation of the old index

### Index Updating

````python
print(f"ser2 is \n{ser2}")
ser2.index = ["D", "B", "A", "C"]
print(f'ser2.index = ["D", "B", "A", "C"] changes the index to uppercase: \n{ser2}')
ser2.index = ser2.index.str.lower()
print(f'ser2.index = ser2.index.str.lower() changes the index back to lowercase by applying the str.lower method: \n{ser2}')
````

### Indexing in Pandas Series

- Allows use of labels in the index for selecting values
    - Assignment of an element works as usual

````python
print(f"ser2 is: \n {ser2}")
print(f'ser2["a"] selects a single value using label: \n{ser2["a"]}')
ser2["d"] = 6
print(f'ser2["d"] = 6 updates the value at label "d": \n{ser2}')
````

### Indexing in Pandas Series

- You can also specify a set of values to be selected
    - Expressed as a list of strings

````python
print(f'ser2[["c", "a", "d"]] selects multiple values using a list of labels: \n{ser2[["c", "a", "d"]]}')
````

### Name Assignment to Series and Indexes

You can  assign a name to the Series and its index to make it more readable


````python
ser2.name = "population"
print(f'ser3.name = "population" assigns a name to the Series: \n{ser2}')
ser2.index.name = "state"
print(f'\ser3.index.name = "state" assigns a name to the index: \n{ser2}')
````

### Operations on Pandas Series

 - Numpy Operations like filtering and Numpy Ufuncs also work on Series
     - The index-value link is unaffected by these operations


````python
print(f'ser2[ser2 > 0] filters values greater than zero: \n{ser2[ser2 > 0]}')
print(f'np.exp(ser2) applies the exponential function: \n{np.exp(ser2)}')
````

### DataFrame in Pandas

Pandas DataFrames:
- Represent rectangular table of data
    - Contains ordered, named collection of **columns**
      - A column is a **named Series**
      - Different columns may be of different data types
      - The type of the elements in each column is the same
- Have both **row and column indexes**
    - Row indexes are called `index`
    - Column indexes are called `column`


### DataFrame Creation
- Based on dictionary structure
    - Each dictionary key is a column name
    - Each dictionary value is a column of data
- Values needs to be of equal-length lists, 1D Numpy arrays, Pandas Series 

### DataFrame Creation


````python
dict_data = {
    "state": ["Ohio", "Ohio", "Ohio", "Nevada", "Nevada", "Nevada"],
    "year": [2000, 2001, 2002, 2001, 2002, 2003],
    "pop": [1500000, 1700000, 3600000, 2400000, 2900000, 3200000]
}
df1 = pd.DataFrame(dict_data)
print(f'We want to use the data in dict_data :\ndata = {dict_data}\n')
print(f'Using: df1 = pd.DataFrame(dict_data) we create a DataFrame based on the dictionary: \n{df1}')
````

### Pandas Data Types

- Pandas DataFrames are versatile and can hold a variety of data types (most of Numpy and more)
- Each column in a DataFrame can have its own data type
  - Use `df.dtypes` to view the data type of each column in a DataFrame
  - Use `df['column_name'].dtype` to view the data type of a single column in a DataFrame
  - Use `astype()` to convert to a given data type (the usual caveats/warnings apply)
  


### DataFrame Index and Column Ordering

- Columns are ordered based on the key order in the dictionary
    - You can order the columns by specifying it to the `columns` parameter
    - If you pass a column that isn't in the dictionary, it will appear with missing values
- The index is automatically assigned, similar to Series
    - The DataFrame `df` will have its index starting from 0 and incrementing by 1
    - You can specify the index by passing a list to the `index` parameter or modify it using the `index` method





### DataFrame Index and Column Ordering

````python
df2 = pd.DataFrame(dict_data, columns=["year", "state", "pop", "debt"])
print(f'''df2 = pd.DataFrame(dict_data, columns=["year", "state", "pop", "debt"]) constructs a 
DataFrame with specified column order and an additional "debt" column (with all NaN since no data about it is available): \n{df2}\n''')
df2.index = ['a', 'b', 'c', 'd', 'e', 'f']# Changing the index to string labels
print(f'The command df2.index = ["a", "b", "c", "d", "e", "f"] changes the index to string labels: \n{df2}')
````

### DataFrame Using a Column as Index - The `set_index` Method

- You can use the `set_index` method to create a new index based on a column 
    - The column will be removed from the DataFrame
- This creates a new DataFrame object, does not modify the original one


````python
print(f"The command df2.set_index('year') creates a new dataFrame using 'year' as the index and removes it from the columns: \n{df2.set_index('year')}\n")
print(f"df2 is still \n{df2}\n")
````

### DataFrame Using a Column as Index - The `set_index` Method - The `drop` Parameter

- The index is separate from the columns, so it is less convenient to access
    - Some variables are useful to use both as index and for computations 
    - Time is often needed to identify observations, but also needed to compute obs age or distance from an event
- Use the `drop` parameter to keep the column in the DataFrame


````python
print(f"The command df2.set_index('year', drop=False)) creates a new dataFrame using 'state' as the index and removes it from the columns: \n{df2.set_index('year', drop=False)}\n")
````

### DataFrame Using a Column as Index - The `set_index` Method - The `inplace` Parameter

- Using the parameter `inplace=True` will modify the DataFrame in place
    - **This is a functionality available in many Pandas methods**

````python
df2.set_index('year', drop=False, inplace=True)
print(f"The command df2.set_index('year', drop=False, inplace=True) updates df2 setting 'year' as the index and keeps it among the columns: \n{df2}\n")
````

### DataFrame Using a Column as Index - The `reset_index` Method

- The `reset_index` method does the opposite of set_index
    - It takes the index and makes it a column in the DataFrame
- The index is then reset to the default integer index

````python
print(f"The command df2.set_index('state').reset_index() retrieves 'state' back from the index into the DataFrame: \n{df2.set_index('year').reset_index()}\n")
df2.index = ['a', 'b', 'c', 'd', 'e', 'f']
print(f"""Using df2.index = ['a', 'b', 'c', 'd', 'e', 'f'] we get to the original form of df2 : \n{df2}\n""")
````

### Using Hierarchical Indexing for Multi-Dimensional Data

- Pandas dataFrames are inherently two-dimensional
    - Usually in a tidy dataset we:
        - Store different measures - in the column dimension
        - For different observations - in the row dimension 
- In real-world scenarios, data can be three-dimensional or higher:
    - E.g., panel data contains three dimensions: observations (row), measures (column), and time (?) 
        - Numpy address this problem with 3D arrays (from MatLab) 
- Hierarchical indexing provides a way to represent multi-dimensional data within the two-dimensional structure of a DataFrame
    - This feature allows to have multiple index levels on an axis
    - Capturing the complexity of panel data or higher-dimensional data

### Defining a MultiIndex on Existing DataFrames

- Just pass to the `set_index` method multiple column names

````python
print(f"df1 is: \n{df1}\n")
df1.set_index(['state', 'year'], inplace=True)
print(f"The command df1.set_index(['state', 'year'], inplace=True) set a MultiIndex (state, year): \n{df1}\n")
````

### Column Creation in Pandas DataFrame - Direct Assignment Using a Scalar Value


````python
df2["debt"] = 2e9
print(f'df2["debt"] = 2e9 adds a new column "debt" with the scalar of 2 billions: \n{df2}')
df2["country"] = "USA"
print(f'df2["country"] = "USA" adds a new column "country" with a string value "USA": \n{df2}')
````

### Column Creation in Pandas DataFrame - Direct Assignment Using Lists or Arrays

- Using Lists or Arrays: Use a list to populate a new column
    - The list length must match the DataFrame length

````python
# Creating a list of values
debt_values = [1.2e9, 1.5e9, 1.7e9, 0.5e9, 0.8e9, 1.0e9]
print(f'debt_values = [1.2e9, 1.5e9, 1.7e9, 0.5e9, 0.8e9, 1.0e9] creates a list of values: \n{debt_values}\n')
df2["debt"] = debt_values
print(f'df2["debt"] = debt_values assigns the list to the "debt" column in frame2: \n{df2}')
````

### Column Creation in Pandas DataFrame - Direct Assignment Using Series 

- If you assign a Series to a DataFrame column, index labels are matched to the DataFrame's index
    - The Series index takes precedence on order in alignment
    - Any DataFrame index not present in the Series will have `NaN` in the new column

### Column Creation in Pandas DataFrame - Direct Assignment Using Series - Matching Indexes


````python
# Creating a new Series with specified index that matches the DataFrame's index
income_series = pd.Series([50.2e3, 45.3e3, 60.1e3, 55.5e3, 42.8e3, 38.9e3], index=['b', 'c', 'd', 'e', 'f', 'a'])
print(f"""income_series = pd.Series([50.2e3, 45.3e3, 60.1e3, 55.5e3, 42.8e3, 38.9e3], index=['b', 'c', 'd', 'e', 'f', 'a']) 
creates a new Series with specified index:\n{income_series}\n""")
df2["income"] = income_series # Assigning the Series to a DataFrame column
print(f'df2["income"] = income_series assigns the Series to the "income" column in df2, aligning by index: \n{df2}')
````

### Column Creation in Pandas DataFrame - Direct Assignment Using Series - Non-Matching Indexes


````python
# Creating a new Series with specified index (not matching the one of the dataFrame)
val = pd.Series([2e3, 3e3, 4e3], index=['b', 'd', 'f'])
print(f"val = pd.Series([2e3, 3e3, 4e3], index=['b', 'd', 'f'] creates a new Series with specified index:\n{val}\n")
df2["income"] = val # Assigning the Series to a DataFrame column - notice that we overwrite ALL the previous values
print(f'df2["income"] = val assigns the Series to the "debt" column in frame2, aligning by index:\n{df2}')
````

### Column Creation in Pandas DataFrame - Derived Columns

### Row Retrieval in Pandas DataFrame - `loc` and `iloc`

- Rows can be retrieved using `loc` and `iloc` methods
  - `loc` is label-based data selection
      - Need to specify the index label of the row to select
  - `iloc` is  position-based data selection
      - Need to specify the position of the row to select

````python
print(f"df2 is: {df2}")
````

### Row Retrieval in Pandas DataFrame - `loc`

- `loc` is label-based data selection
  - Need to specify the index label of the row to select

````python
print(f'df2.loc["b"] retrieves the row with index label "b": \n{df2.loc["b"]}\n')
print(f'df2.loc[["b", "c"]] retrieves the row with indexes label "b" and "c": \n{df2.loc[["b", "c"]]}')
````

### Row Retrieval in Pandas DataFrame - `iloc`

- `iloc` is  index-based data selection
  - Need to specify the position of the row to select

````python
print(f'df2.iloc[1] retrieves the row with integer position 2, which corresponds to label "b": \n{df2.iloc[1]}\n')
print(f'df2.iloc[[1, 2]] retrieves the row with integer position 2 and 3, which correspond to label "b" and "c": \n{df2.iloc[[1, 2]]}')
````

### Conditional Selection in DataFrames

- Conditionals are used to filter rows based on specific criteria, same as Numpy
    - The conditional expression returns a Boolean Series

````python
# Creating a Boolean series values by comparing each column element with a scalar
df2['year'] > 2001
print(f"df2['year'] > 2001 returns a Boolean Series: \n{df2['year'] > 2001}")
````

### Conditional Selection in DataFrames

- Conditionals are used to filter rows based on specific criteria, same as Numpy
    - The Boolean Series used in indexing selects rows with True and drops the False ones

````python
# Selecting rows where the year is greater than 2001
print(f"df2[df2['year'] > 2001] selects rows where the 'year' column is greater than 2001: \n{df2[df2['year'] > 2001]}")
````

### Using Boolean Arrays with `loc`

- Boolean arrays can be used with `loc`
    - This feature is not available with `iloc`

### Combining Row and Column Selection in Pandas

- You can combine both row and column selection
  - In `loc`, you separate the label-based indexes of row and column with a comma
  - In `iloc`, you separate the  integer-based indexing with a comma


### Combining Row and Column Selection in Pandas

````python
print(f'df2 is: \n{df2}\n')
print(f'''df2.loc[["a", "b"], ["year", "state", "pop"]] retrieves specific rows and columns by label:
{df2.loc[["a", "b"], ["year", "state", "pop"]]}\n''')
print(f'df2.iloc[[0, 1], [0, 1, 2]] retrieves specific rows and columns by index: \n{df2.iloc[[0, 1], [0, 1, 2]]}\n')
````

### Integer Indexing: Label-based vs Position-based

- Do not use `df[row_mask, col_mask]` without `.loc/.iloc`, you will get errors

- Chained indexing like `df[rows][cols]` can work for reading but is unsafe 

- In Pandas, using integers for indexing can be ambiguous
    - The cases where is unclear if label-based or position-based indexing should be applied
- Pandas throws an error when there are ambiguities
- It is recommended to **be explicit using `loc` and `iloc`** so to avoid confusion


### Dropping Columns and Rows in Pandas - The `drop` Method
- In a data analysis workflow, you often need to drop rows and columns
    - Observations not of interest 
    - Intermediate columns no longer needed
- The `drop` returns a new DataFrame without the selected rows/columns
    - The original DataFrame is not modified, but can be using `inplace=True`
- `drop` accepts various parameters, my advice is to use:
    - `columns`:  column name(s) to drop
    - `index`: Index label(s) to drop

### Dropping Columns and Rows in Pandas - Dropping a Column

````python
print(f'df2 is: \n{df2}\n')
print(f'df2.drop(columns="per_capita_debt") drops the "per_capita_debt" column: \n{df2.drop(columns="per_capita_debt")}')
````

### Dropping Columns and Rows in Pandas - Dropping a Column in Place

````python
print(f'df2 is still: \n{df2}\n')
df2.drop(columns=["country", "per_capita_debt"], inplace=True)
print(f'df2.drop(columns=["country", "per_capita_debt"], inplace=True) removes from df the the "country" and "per_capita_debt" columns: \n{df2}')
````

### Dropping Columns and Rows in Pandas - Dropping a Row

````python
print(f'df2 is: \n{df2}\n')
print(f'df2.drop(index="a") drops the row with label "a": \n{df2.drop(index="a")}')
````

### Sorting
- Sorted data allows to quickly identify patterns along with potential outliers/errors/missing values

        - You want to see how it varies across (groups of) observations
            - E.g., are there more females or males at the extremes?
        - Are there inexplicably tall or short children? have there been recording errors?
- Ranking is just a special case of sorting
    - Assigns ranks from one through the number of valid data points in an array
    - The rank of a value is its index in the sorted array


### Sorting to Present Analysis' Results

- Apart from easen data exploration, sorting plays a major role in presenting analysis results
- One of the most common way of presenting analysis results is by means of tables
    - Pandas Dataframe are already in a tabular format
        - That are rendered as fairly nice tables in Jupyter Notebooks
        - Can also be exported in a variety of tabular formats (e.g., excel)
- A **CRUCIAL** aspect of a good presentation of result is its **clarity**
    - The information that you want to provide should be easily accessible to the reader
- Sorting is the most basic prerequisite to make a table clear and easy to read

### Sorting in Pandas
- Sorting in Pandas can be done using the `sort_index` and `sort_values` methods
- `sort_index` sorts the data lexicographically by index or column label
    - By default, sorting is done in ascending order. Use `ascending=False` for descending
- `sort_values` sorts a pandas object by its values
    - Ascending takes a Boolean list to specify sorting order for each column
- Both methods return a new, sorted object
    - The original object is not modified
    - Use `inplace=True` to modify the original object

````python
print(f"ser2 series is: \n{ser2}\n")
print(f"ser2.sort_index() sorts by index: \n{ser2.sort_index()}\n")
print(f"ser2.sort_index(ascending=False) sorts by index in descending order: \n{ser2.sort_index(ascending=False)}\n")
````

````python
print(f"ser2.sort_values() sorts by values: \n{ser2.sort_values()}\n")
print(f"ser2.sort_values(ascending=False) sorts by values in descending order: \n{ser2.sort_values(ascending=False)}\n")
````

````python
print(f"ser2 has not been modified by previous reordering: \n{ser2}\n")
ser2.sort_index(ascending=False, inplace=True) 
print(f"ser2.sort_index(ascending=False, inplace=True) ser2 gets changed to: \n{ser2}\n")
````

## Dealing with Missing Values and Duplicated Observations in Pandas

### Missing Data 

- Missing data is a common issue in data analysis 
- Consists of **non-existent** or **unobserved** data
- It is important to have a clear picture of missing data in the dataset under analysis:
  - Provides info of the representativeness of the data
  - Identifies potential biases that might invalidate the analysis
  - Might point towards data collection problems  

### Missing Data in Pandas - Values and Methods

Pandas aims to simplify dealing with missing data
- There are two objects used as **sentinel values** for missing data:
    - `NaN` (Not a Number, from Numpy) to represent missing data for numeric values
    - `None`(from basic Python) to represent other missing data
    - This distinction is important for some operations, but these are mainly edge cases
- Pandas provides methods to handle missing data:
    - `isna` (`notna`) returns a Boolean Series indicating which values are (not) missing

### Identifying and Removing Missing Data


````python
ser_nan = pd.Series([1.2, np.nan, None,  -3.5, 2, 0], index=["a", "b", np.nan, "d", np.nan, "f"])
print(f"The series ser_nan is:\n{ser_nan}\n")
print(f"The command ser_nan.isna() returns a boolean Series with True for NaN values: \n{ser_nan.isna()}\n")
print(f"The command ser_nan.isna().sum() easily computes the number of NaN values: \n{ser_nan.isna().sum()}\n")
````

### Filtering Missing Data in Series
- To Filter Missing Data:
    - Manual filtering using `isna`/`notna` and Boolean indexing
    - Automated approach using `dropna` method
- `dropna` on a Series only return elements that have non-missing value
- The `inplace` parameter allows to modify the Series in place

### Filtering Missing Data in Series


````python
print(f"The series ser_nan is:\n{ser_nan}\n")
print(f"The command ser_nan[ser_nan.notna()] returns it with NaN values removed: \n{ser_nan[ser_nan.notna()]}\n")
print(f"The command ser_nan.dropna() returns it with NaN values removed: \n{ser_nan.dropna()}\n")
````

### Filtering Missing Data in DataFrames
- Removing missing data in dataFrames offers more options:
    - `axis={0,1}` sets whether to drop rows or columns
    - `how={any, all}` sets whether to drop a row/column that have any/all  NA
    - `thresh` sets the minimum number of non-NA values required to keep a row/column

### Filtering Missing Data in DataFrames

````python
df_nan = pd.DataFrame({'A': [4, 1, 3, np.nan],'B': [3, np.nan, 2, np.nan],'C': [3, np.nan, 1, np.nan],'D': [1, 2, np.nan, np.nan]})
print(f"DataFrame df_nan:\n{df_nan}\n")
print(f"The command df_nan.dropna() drops all rows with any NaN values: \n{df_nan.dropna()}\n")
````

### Filtering Missing Data in DataFrames

````python
print(f"DataFrame df_nan:\n{df_nan}\n")
print(f"df_nan.dropna(axis=0, how='all') drops all rows with all NaN values: \n{df_nan.dropna(axis=0, how='all')}\n")
print(f"df_nan.dropna(axis=1, thresh=3) drops all columns with less than three non-NaN values: \n{df_nan.dropna(axis=1, thresh=3)}\n")
````

### Duplicate Data in Pandas

- Duplicate data is a common issue in data analysis
    - Consists of **repeated** or **identical** data
    - Notice that in duplicates both index and values are identical
- Pandas includes methods to easily detect and handle duplicate data
    - `duplicated` returns a Boolean Series indicating which rows are duplicates
    - `drop_duplicates` returns a DataFrame with duplicate rows removed
        - `drop_duplicates` has a `subset` parameter to specify which columns to consider for duplicates

### Identifying and Removing Duplicates

````python
df_dup = pd.DataFrame({"col1": ["a", "b", "a", "a", "b"],  "col2": [1, 1, 2, 1, 1]})
print(f"DataFrame df_dup:\n{df_dup}\n")
print(f"df_dup.duplicated() returns a boolean Series with True for duplicated rows: \n{df_dup.duplicated()}\n")
print(f"df_dup.drop_duplicates() drops duplicated rows: \n{df_dup.drop_duplicates()}\n")
````

### Identifying and Removing Duplicates Based on a Subset of Columns

````python
df2_dup = pd.DataFrame({"col1": ["a", "b", "a", "a", "b"],  "col2": [1, 1, 2, 1, 1], "col3": np.arange(5)})
print(f"DataFrame df_dup:\n{df2_dup}\n")
print(f"The command df2_dup.drop_duplicates() drops duplicated rows: \n{df2_dup.drop_duplicates()}\n")
print(f"The command df2_dup.drop_duplicates(subset=['col1', 'col2']) drops rows if the columns ['col1', 'col2'] are duplicated: \n{df2_dup.drop_duplicates(subset=['col1', 'col2'])}\n")
````

### Duplicated Index Labels in Pandas

- Using duplicate index labels it's generally not recommended as data selection becomes messy
    - Indexing a label with multiple entries returns a Series
    - Indexing a label with a single entry returns a scalar value
- Usually happens because of a mis-specified multi-index
    - e.g., it shoud be obs-year but it is only obs
- **If you are getting unexpected results, check if your labels are unique**
    - The `is_unique` property can be used to check if axis labels are unique

````python
ser7 = pd.Series(np.arange(5), index=["a", "a", "b", "b", "c"])
print(f"Series with duplicate index labels: \n{ser7}\n")
print(f"ser7.index.is_unique checks for unique index labels and returns: {ser7.index.is_unique}\n")
````

## Iteration and Vectorization in Pandas

### Iterating over Pandas DataFrames

- Iterating over Pandas DataFrames is **not recommended**
    - Pandas is optimized for vectorized operations - **iterating over rows is slow**
    - Still, iteration is the most flexible tool, so sometimes it is necessary
- You can iterate using any index method (bracket, `loc`, `iloc`)
    - The index method (`df.index`) returns a Series with the indexes for each row
    - You can use these indexes to select each row in a loop
- Iterrows is a specificaly designed method for iterating over rows
    - It returns a tuple of two elements containing the **index** and **row** values

### DataFrame's Apply Method in Pandas

- The `apply` method in Pandas is the 'intended' way to apply functions to a DataFrame
- Not needed for array statistics commonly used(e.g., `sum` and `mean`)
    - `df.sum()` and `df.mean()` are equivalent to `df.apply(sum)` and `df.apply(mean)`
- Most useful with **user-defined functions**

### Advantages of Using Functions

- A function is a "chunk" of code that takes inputs (parameters) and provides an output (return value)
- In principle, it could be substituted by copy-pasting the code. **However functions allow for**:
   1. **Evocative Naming**: Can be assigned an evocative name, making the code easier to understand
   2. **Centralized Code Updates**: Only need to update code in one place when needed
   3. **Reduced Error Risk**: Eliminate the chance of incidental mistakes when copy-pasting
   4. **Exploit Pandas's `apply`**: Operate on the elements of multiple rows/columns without iteration
- **Best Practices**:
  - Functions should do **only one thing** and do it well
  - Create a function if you find yourself copying and pasting the same code more than twice

### Defining a Function in Python

```python
def function_name(parameters, separated, by, commas):
    """docstring""" # to describe what the functions do - optional but advised
    body_doing_operations_on_parameters_and_storing_result
    return result
```
  - `def`: Keyword to start a function definition
  - `function_name`: Unique identifier for the function
  - `parameters`: Variables to pass to the function
  - `docstring`: Optional, for documentation of what the function does
  - `body`: Code block where the function's tasks are performed
  - `return`: Keyword to return a value from the function
      - if not specified the last line of the body is returned

### An Example of Function Definition in Python

- **Function `pass_or_fail`**:
  - input: `score` on the test
  - output: `"Pass"`  if the score is above the `passing_threshold` and `"Fail"` otherwise

````python
def pass_or_fail(score, passing_threshold):
    """Determines if the score is a pass or fail based on the passing threshold.""" #Docstring describing the function
    if score < 0 or score > 100:                                                    #Check if the score has an admissible value
        return "Invalid score"                                                      #Of the score value is not admissible return "Invalid score"
    if score >= passing_threshold:                                                  #Check if the the score is bigger or equal to the passing_threshold
        return "PASSED"                                                             #If the the score is bigger or equal to the passing_threshold return "Passed"
    else:                                                                           #Otherwise
        return "FAILED"                                                             #Return "FAILED"
print(f"pass_or_fail(score=85, passing_threshold=60) returns: {pass_or_fail(score=85, passing_threshold=60)}")
print(f"pass_or_fail(score=50, passing_threshold=60) returns: {pass_or_fail(score=50, passing_threshold=60)}")
print(f"pass_or_fail(score=110, passing_threshold=60) returns: {pass_or_fail(score=110, passing_threshold=60)}")
````

### `apply` a Custom Function on a DataFrame -  Using Multiple Columns
- `apply` allows to define functions that combine multiple columns
- Let's define a function that returns the ratio of two columns

````python
print(f"The function calculate_debt_to_income(row), calculates the ratio of the columns 'debt' and 'income'")
def calculate_debt_to_income(row):
    if row['income'] != 0:
        return row['debt'] / row['income']
    else:
        return np.nan
````

## Group Operations in Pandas

### Group Operations

- Group operations:
    - **Split** a dataset into groups
    - **Apply** a function to each group independently
    - **Combine** the results into a data structure
- Computing how a measure varies across groups:
    - Allow to characterize the measure (its distribution, its mean) for each group
        - E.g., how the height of children varies across schools
    - Allows to see how the measure varies across groups $\rightarrow$ investigate its relationship with the group variable
        - E.g., how the height of children varies with exposure to pollutants

### Understanding the Split-Apply-Combine Strategy in pandas

- **A Three-Stage Process**:
  1. **Splitting**:
     - Dividing the data (Pandas dataFrame) into groups based on one or more keys
  2. **Applying**:
     - Performing a function on each group
     - This will create a new object that might be: 
         - Scalar (resulting from an aggregation, e.g., count, sum) 
         - Composite (e.g., dataFrame or Series)
  3. **Combining**:
     - Aggregating the results from each group into a single result object
        - The structure of the result depends on the operation performed

### Grouping Keys 

- Grouping keys can vary in type:
  1. **Column Names**: Of the DataFrame to be grouped (most common)
  2. **List/Array**: Directly specifying the groups (the length must match with the axis to group)
  3. **Dictionary/Series**: Groups together obs whose index is associated to the same value in the Dictionary/Series
  4. **Index Levels**: Groups together obs with the same value in a given level of a multiindex
  5. **Function**: Groups together obs with the same `function(index)` result

### Data Aggregation Using `apply` - An Example

    - We want to identify the highest n top and bottom tips in percentage by groups based on `day`/`time` combination


### Data Aggregation Using `apply` - The Steps of the Example
1. Data Import
2. Definition of a function `compute_pctTip_topBot`
    1. Computes the tip percentage of a bill 
    2. Select the highest n top and bottom tips in percentage
3. Application of the function by groups based on `day`/`time` combination
4. Refinements to improve readability (**CRUCIAL**!) of the table result
5. Saving the result to a file

### Importing the `tips` Dataset

````python
import os
#reads the file tips.csv and stores it into a df
tips = pd.read_csv('data/tips.csv')
print(f"Using pd.read_csv('data/tips.csv') to read tips.csv and store the data in the variable tips:")
tips
````

### Some Info on the `tips` Dataset
The dataset `tips` contains information about tips in a restaurant:
- `total_bill`: Total bill (cost of the meal), including tax, in US dollars
- `tip`: Tip amount in US dollars
- `sex`: Gender of the person attending the table
- `smoker`: Indicates whether the table was in the smoking area or not
- `day`: Day of the week
- `time`: Meal (e.g., Lunch, Dinner)
- `size`: Number of people in the party

### Grouping by `day`/`time` Combination
- The dataset provides the keys that we need for grouping in the columns `day` and `time`
- We need to define the function to apply to each group and extract the information that we need

### Function Definition - `compute_pctTip_topBot`

````python
# Defining the function compute_ratio_topBottom
def compute_pctTip_topBot(group, n):
    # Calculate the percentage of tip from the total bill and add it as a new column 'pctTip'
    group['pctTip'] = 100 * group['tip'] / group['total_bill']
    # Select the top n rows with the highest 'pctTip' values
    top_n = group.nlargest(n, 'pctTip')
    # Select the bottom n rows with the lowest 'pctTip' values
    bottom_n = group.nsmallest(n, 'pctTip')
    # Concatenate the top and bottom rows into a single DataFrame
    #more on this soon, concatenate stacks the two dataframes one on top of the other
    result = pd.concat([top_n, bottom_n]) 
    # Return the combined DataFrame
    return result
````

### Apply the Function `compute_pctTip_topBot` by `day`/`time` Groups

````python
# Grouping the 'tips' DataFrame by 'day' and 'time', and applying the function to each group
tips.groupby(['day', 'time']).apply(compute_pctTip_topBot, n=2, include_groups=False)
````

### Refinements to the Table to Improve Readability

````python
# Map the weekday names to their corresponding numerical values
weekday_map = {'Mon': 1, 'Tue': 2, 'Wed': 3, 'Thur': 4, 'Fri': 5, 'Sat': 6, 'Sun': 7}
# Map the meal times (e.g., 'Lunch', 'Dinner') to numerical values for sorting
time_map = {'Lunch': 1, 'Dinner': 2}
# Apply the weekday mapping to the 'day' column, creating a new column 'day_order'
tips['day_order'] = tips['day'].map(weekday_map)
# Apply the time mapping to the 'time' column, creating a new column 'time_order'
tips['time_order'] = tips['time'].map(time_map)
tips
````

### Refinements to the Table to Improve Readability

````python
# Group the DataFrame by 'day' and 'time', apply the compute_ratio_topBottom function to each group
# Sort the resulting DataFrame by 'day_order' and 'time_order'
df_pctTip_topBot_by_dayTime = tips.groupby(['day', 'time']).apply(compute_pctTip_topBot, n=2,include_groups=False)\
                  .sort_values(['day_order', 'time_order'])
df_pctTip_topBot_by_dayTime
````

### Refinements to the Table to Improve Readability

````python
# Group the DataFrame by 'day' and 'time', apply the compute_ratio_topBottom function to each group
# Sort the resulting DataFrame by 'day_order' and 'time_order'
# Drop the columns 'day', 'day_order', 'time', and 'time_order' from the final result
df_pctTip_topBot_by_dayTime = df_pctTip_topBot_by_dayTime.drop(columns=["day_order", "time_order"])\
                  .reset_index(level=[2], drop=True) #From the MultiIndex result (day/time/row), drop the second level (row)
df_pctTip_topBot_by_dayTime
````

### Refinements to the Table to Improve Readability

````python
# For some day/time combinations, there are less than 4 observations leading to duplicates, we remove them Drop the columns 'day', 'day_order', 'time', and 'time_order' from the final result
df_pctTip_topBot_by_dayTime = df_pctTip_topBot_by_dayTime.drop_duplicates()
df_pctTip_topBot_by_dayTime
````

### Arithmetic with Pandas - Automatic Index Alignment

- Pandas simplifies doing operations on multiple objects by accounting for their indexes
- When performing arithmetic operations and other data transformations, Pandas automatically aligns indexes
    - **The resulting index is the union of the indexes of the two Pandas objects**
    - If index pairs don't match, missing values (`NaN`) are introduced


### Arithmetic with Pandas - Automatic Index Alignment in Series


````python
ser3 = pd.Series([7.3, -2.9, 3.4], index=["a", "c", "d"])
ser4 = pd.Series([-2.1, 2.7, -1.5], index=["c", "a", "e",])
print(f"ser3 Series: \n{ser3}\n")
print(f"ser4 Series: \n{ser4}\n")
print(f"Result of ser3 + ser4: \n{ser3 + ser4}")
````

### Arithmetic with Pandas - Automatic Index Alignment in Series

- `ser3 + ser4` yields a series with indexes `a, c, d, e`
    - Overlapping indexes (`a` and `c`) are matched in the operation 
    - Non-overlapping indexes (`d` and `e`) return `NaN`

### Arithmetic with Pandas - Automatic Index Alignment in DataFrames

- The default behavior of Pandas is to **align series indexes with the dataFrame columns**
    - The resulting index is the union of the indexes of the two Pandas objects
    - If index pairs don't match, missing values (`NaN`) are introduced (a union of the indexes)

### Arithmetic with Pandas - Automatic Index Alignment in DataFrames


````python
df4 = pd.DataFrame(np.arange(9.).reshape((3, 3)).transpose(), columns=["Colorado", "Texas", "Ohio"], index=["Colorado", "Texas", "Ohio"])
print(f"DataFrame df4:\n{df4}\n")
ser5 = pd.Series([10, 20, 30], index=["Colorado", "Texas", "Florida"])
print(f"Series ser5:\n{ser5}\n")
print(f"df4 + ser5 aligns the Series index with the DataFrame columns, broadcasting down the rows: \n{df4 + ser5}\n")
````

### Arithmetic with Pandas - Automatic Index Alignment in DataFrames

- With Pandas arithmentic methods (e.g., `add`), use `axis=0` to align with the rows

````python
print(f"DataFrame df4:\n{df4}\n")
print(f"Series ser5:\n{ser5}\n")
print(f"df4.add(ser5, axis=0) aligns the Series index with the DataFrame rows, broadcasting across the columns: \n{df4.add(ser5, axis=0)}\n")
````

### Index Alignment for Arithmetic Operations

### Arithmetic with Pandas - Automatic Index Alignment

- Pandas simplifies doing operations on multiple objects by accounting for their indexes
- When performing arithmetic operations and other data transformations, Pandas automatically aligns indexes
    - **The resulting index is the union of the indexes of the two Pandas objects**
    - If index pairs don't match, missing values (`NaN`) are introduced


### Arithmetic with Pandas - Automatic Index Alignment in Series


````python
ser3 = pd.Series([7.3, -2.9, 3.4], index=["a", "c", "d"])
ser4 = pd.Series([-2.1, 2.7, -1.5], index=["c", "a", "e",])
print(f"ser3 Series: \n{ser3}\n")
print(f"ser4 Series: \n{ser4}\n")
print(f"Result of ser3 + ser4: \n{ser3 + ser4}")
````

### Arithmetic with Pandas - Automatic Index Alignment in Series

- `ser3 + ser4` yields a series with indexes `a, c, d, e`
    - Overlapping indexes (`a` and `c`) are matched in the operation 
    - Non-overlapping indexes (`d` and `e`) return `NaN`

### Arithmetic with Pandas - Automatic Index Alignment in DataFrames

- The default behavior of Pandas is to **align series indexes with the dataFrame columns**
    - The resulting index is the union of the indexes of the two Pandas objects
    - If index pairs don't match, missing values (`NaN`) are introduced (a union of the indexes)

### Arithmetic with Pandas - Automatic Index Alignment in DataFrames


````python
df4 = pd.DataFrame(np.arange(9.).reshape((3, 3)).transpose(), columns=["Colorado", "Texas", "Ohio"], index=["Colorado", "Texas", "Ohio"])
print(f"DataFrame df4:\n{df4}\n")
ser5 = pd.Series([10, 20, 30], index=["Colorado", "Texas", "Florida"])
print(f"Series ser5:\n{ser5}\n")
print(f"df4 + ser5 aligns the Series index with the DataFrame columns, broadcasting down the rows: \n{df4 + ser5}\n")
````

### Arithmetic with Pandas - Automatic Index Alignment in DataFrames

- With Pandas arithmentic methods (e.g., `add`), use `axis=0` to align with the rows

````python
print(f"DataFrame df4:\n{df4}\n")
print(f"Series ser5:\n{ser5}\n")
print(f"df4.add(ser5, axis=0) aligns the Series index with the DataFrame rows, broadcasting across the columns: \n{df4.add(ser5, axis=0)}\n")
````

### Strategy for Effective Merging - The Checks to Do When Merging
- **Check the keys/ID of each dataset for duplicates**
- **Identify Non-Matching keys/ID Before Merging**:
  - Identify the keys in the receiver dataframe that do not have a match in the donor dataframe
  - For these obs the donated columns will be `NaN` in the merged dataframe
  - Better to use indexes for this check 
    - Allows to use the `.index.get_level_values` method
    - For multiindex DataFrames creates automatically the combinations of indexes

### Strategy for Effective Merging - The Checks to Do When Merging
- **Prefer Adding Columns Individually**:
  - Enhances code readability and manageability
  - Adding one column at a time makes it easier track `NaN` values and perform the checks below:
    1. **Check the NaN in the column to be received** 
        - The column stored in the donor dataframe
        - If there are `NaN` values they will pass to the receiver dataframe
    2. **Check the NaN in the column received** 
        - The column stored in the receiver dataframe (only available after the merge)
        - The number of `NaN` should be equal to the **sum** of:
            - The number of `NaN` in the donor dataframe plus 
            - the number of indexes/keys/ID in the receiver not in the donor
            

### Strategy for Effective Merging - The Datasets

````python
df_WAmod = df_WA[df_WA['ID'] != 'luca'].copy()
print(f"The command df_WAmod = df_WAmod[df_WAmod['ID'] != 'luca'].copy() creates a new DataFrame from df_WA by removing the row with ID 'luca':\n{df_WAmod}\n")
df_HAmod = df_HA[df_HA['ID'] != 'lynne'].copy()
print(f"The command df_HAmod = df_HA[df_HA['ID'] != 'lynne'].copy() creates a new DataFrame from df_HA by removing the row with ID 'lynne':\n{df_HAmod}\n")      
````

### Strategy for Effective Merging - Identifying Non-Matching IDs/indexes/keys
- In a MultiIndex, the `get_level_values()` method returns combinations of the index values at the specified level
    - This uniquely identifies each observation
    - Using a column is fine for a single key
    - With multiple IDs/indexes/keys it is better to use a MultiIndex

### Strategy for Effective Merging - Identifying Non-Matching IDs/indexes/keys


````python
idsReceiver_notIn_donor = set(df_WAmod.index.get_level_values('ID')) - set(df_HAmod.index.get_level_values('ID'))
print(f"""The command
idsReceiver_notIn_donor = set(df_WAmod.index.get_level_values('ID')) - set(df_HAmod.index.get_level_values('ID')) 
identifies unique IDs in df_WAmod (receiver) but not in df_HAmod (donor):\n
{idsReceiver_notIn_donor}""")
````

- This check is in line with our expectations
    - No NaN arising due to typing differences or oddities due to duplicates
- It identifiee the IDs that will result in NaN during the merge

### Strategy for Effective Merging - Compare the NaN Pre/Post Merge
- We want to check the indexes of the NaN in:
    1. The column **to be received** (in the donor DataFrame) 
    2. The column **effectively received** after merge (in the receiver DataFrame)

````python
heightNan_idsDonor = set(df_HAmod[df_HAmod['height'].isna()].index.get_level_values('ID'))
print(f"""The command
heightNan_idsDonor = set(df_HAmod[df_HAmod['height'].isna()].index.get_level_values('ID')) 
stores the index of the NaN values in the 'height' column of the donor: \n{heightNan_idsDonor}\n""")
heightNan_idsReceived = set(df_WAmodJoined[df_WAmodJoined['height'].isna()].index.get_level_values('ID'))
print(f"""The command
heightNan_idsReceived = set(df_WAmodJoined[df_WAmodJoined['height'].isna()].index.get_level_values('ID'))
stores the indexes of the NaN values in the 'height' column received: \n{heightNan_idsReceived}""")
````
