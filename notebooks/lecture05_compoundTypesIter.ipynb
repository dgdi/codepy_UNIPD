### Lists

Lists are defined using square brackets with elements separated by commas:

    [ element_at_index_0, ..., element_at_index_n ]
- A list allows to store multiple elements
    - `element_at_index_0` and `element_at_index_n`.

### Creating Lists Using the Asterisk (*) and Plus (+) Operators in Python

**Bullet Points:**
- The asterisk `*` operator can be used to create lists with repeated elements
    - Syntax for asterisk: `[element] * n`, where `element` is the item to repeat and `n` is the number of repetitions
- The plus `+` operator is used for list concatenation.
    - Syntax for plus: `[list1] + [list2]`, where `list1` and `list2` are the lists to concatenate
- Both operators can be combined for more complex list creations
- Useful for initializing lists, extending lists, or creating patterned lists

````python
# Create a list of zeros with length 5 using *
zeros_list = [0] * 5
print(f"List of zeros with length 5 using *: \n{zeros_list}")
# Create a list combining * and + operators
mixed_list = [1, 2, 3] + [0] * 2
print(f"List combining * and + operators: \n{mixed_list}")
# Extend a list using + operator
extended_list = ["Hi", "Ciao", "Ola"] + [4, 5, 6]
print(f"Extended list using + operator: \n{extended_list}")
# Create a 2D list (list of lists) using * operator
two_d_list = [[0] * 3] * 2
print(f"2D list using * operator: \n{two_d_list}")
````

### Lists are Mutable

So we can use Indexing, slicing and striding to update elements


### Update an Element via Indexing

We can update a single element at a given index as in the following example.

````python
slides_list = [ 'Harry', 'Hermione', 'Nevil', 'Malfoy', 'Sirius' ]
print(f'Before updating the list is: {slides_list}')

# update the element at index 0
slides_list[0] = 'Ronald' 
print(f'After updating the list is: {slides_list}')
````

### Update Several Elements via Slicing
- We can update several elements at once using slicing
- From a given index (included) up to another (excluded)


````python
slides_list = [ 'Harry', 'Hermione', 'Nevil', 'Malfoy', 'Sirius' ]
print(f'Before updating the list is:\n{slides_list}')

# update the elements at index 1 and 2
slides_list[ 1 : 3 ] = [ 'Ginny', 'Luna' ]
print(f"\nAfter updating the list using the command:\nslides_list[ 1 : 3 ] = [ 'Ginny', 'Luna' ]\nthe list becomes:\n{slides_list}")
````

### Update Several Elements via Slicing and Striding

- We can selectively update via slicing and striding
- From a given index (included) up to another (excluded) in steps


````python
slides_list = [ 'Harry', 'Hermione', 'Nevil', 'Malfoy', 'Sirius', 'Gary' ]
print(f"Before updating the list is:\n{slides_list}")

# update all list elements at even indexes 
slides_list[::2] = [ 'Ginny', 'Luna', 'Chris' ]
print(f"\nAfter updating the list using the command:\nslides_list[::2] = [ 'Ginny', 'Luna', 'Chris' ]\nthe list becomes:\n{slides_list}")
````

### Delete Several Elements

We can delete several elements at once by assigning the empty list to a slice


````python
slides_list = [ 'Harry', 'Hermione', 'Nevil', 'Malfoy', 'Sirius' ]
print(f"Before deleting the list is:\n{slides_list}")# remove from index 1 (included) up to 3 (excluded)
slides_list[1:3] = []
print(f"\nAfter deleting elements using the command:\nslides_list[1:3] = []\nthe list becomes:\n{slides_list}")
slides_list[:] = []# empty the list
print(f"\nAfter emptying the list using the command:\nslides_list[:] = []\nthe list becomes:\n{slides_list}")
````

### List Methods - Append, Insert and Remove

* `L.append(e)` adds the element `e` to the end of the list `L`
* `L.insert(i, e)` inserts the element `e` at index `i` in the list `L`
* `L.remove(e)` removes the first occurrence of `e` from the list `L`

### List Methods - Append, Insert and Remove


````python
# Demonstrating the list methods with examples
L = [3, 1, 4, 1, 5, 9]
print(f"Initial list: {L}\n")
# Using append
L.append(2)
print(f"After appending 2: {L}\n")
# Using insert
L.insert(2, 6)
print(f"After inserting 6 at index 2: {L}\n")
# Using remove
L.remove(1)  # removing the first occurrence of 1
print(f"After removing the first occurrence of 1: {L}\n")
#using pop - remove and display based on index 
L.pop(3)
print(f"After removing the fourth element: {L}")
````

### List Methods - Extend and Sort

* `L.sort()` sorts the elements of list `L` in ascending order
    - If the elements are **numbers**, `.sort()` uses the standard order among numbers. 
    - If the elements are **strings**, `.sort()` uses the lexicographical order (the order used in dictionaries) and upper-case letters precede lower-case letters 

* `L.extend(L1)` adds the elements of list `L1` to the end of the list `L` (notice the difference with append)

For a complete list of methods, see [here](http://docs.python.org/3/library/stdtypes.html#mutable-sequence-types)

### List Methods - Sort


````python
# Using sort
L.sort()
print(f"After sorting: {L}\n")

# List of strings
L_string_notSort = ['c', 'a', 'b', 'C', 'A', 'B']

# Store the unsorted list before sorting
# This is needed because .sort(), and methods in general, tend to modify the original data
L_string_sortMethod = L_string_notSort.copy()
# Sorting the list
L_string_sortMethod.sort()

# There are usually functions that are analogous to methods that directly return a copy
L_string_sortFunc = sorted(L_string_notSort)

# Showing both the unordered and ordered versions
print(f"The unordered list of strings: {L_string_notSort}\n")
print(f"The ordered list of strings after sorting using the method .sort():\n{L_string_sortMethod}\n")
print(f"The ordered list of strings after sorting using the function sorted():\n{L_string_sortFunc}")
````

### Function and Methods - Creation of New Variable or In-Place Overwriting

- Usually functions create a new object, while methods modify their object
- E.g., the method `L.sort()` sorts the elements of `L` in place
- If you want to keep the original list, you can use the function `sorted(L)`  
- A very wise policy is to **always** check whether the method you are using is overwriting or not


````python
myList = [4, 5, 2, 1, 3]
print(f"This is myList: {myList}")
myList.sort()  # Sorts in-place. It is more efficient but overwrites the input
print(f"This is mylist after using mylist.sort(): {myList}\n") 

myOtherList = [10, 9, 6, 8, 7]
print(f"This is myOtherList: {myOtherList}")
print(f"this is the result of sorted(myOtherList): {sorted(myOtherList)}")
print(f"This is myOtherList after running sorted(myOtherList): {myOtherList}") 
````

## Iteration


### Iteration

- We will now see how iteration works since it is the prime way to perform operations on lists
     - Its flexibility makes it one of the most important tool in coding at large
- Iteration allows to perform **a set of operations on multiple inputs**
    - Repeating the same code on all the:
        - Elements of a list, columns of a table, different datasets, etc.
- You could simply copy-and-paste the code, but this is:
    - **Sometimes unfeasible** (e.g. if you have a list of 1000 elements)
    - **Error-prone and boring**
    - **Hard to understand and mantain**


### Avoiding Code Duplication Reduces Bugs
- Reducing code duplication has three main benefits:
    1. It's easier to see the **intent** of your code
        - If you have a chunk of code of 20 lines to be applied to 100 elements, it's hard to see what 2000 lines of code are doing
    2. It's easier to respond to **changes** in requirements 
        - e.g. if you have to change the code, you have to change it in 100 places instead of 1
    3. You're likely to have **fewer bugs**
        - Computers do not do mistakes in repetition, humans do


### For Loops in Python
- In Python, loops are defined over iterables
- Iterables are objects that can be iterated over: sequences (lists, dataframes, etc), iterators

### For Loops in Python
```python
for value in iterable:          #Definition of the sequence
    do_something_with_value     #Body
```
- `for value in iterable:` defines the `value` variable that assumes **sequentially** each value in the iterable and executes the code in the body of the loop
    - You can use whatever name instead of `value`
    - Common to use `iterable_loop` 
    - i.e., if the iterable is named `listXXX`, the value name is `listXXX_loop`
- The **body of the loop is indented** and contains the code to be executed for each value in the iterable

### For Loops in Python


````python
print(f"Simple for loop over a list\n") #describe what the code below does
L = [1, 2, 3, 4, 5] #create a list
for val in L: #iterate over the list
    print(f"val = {val}") #print the value of the current iteration
````

### Nested For Loops

- Loops can be combined to create nested loops
- In the simplest setting of two loops there is an **outer loop** and an **inner loop**
- The outer loop is going to iterate across its values
    - For each value on the outer loop the inner loop is going to iterate across its values
- E.g. you have two lists of space (north, south, east, west) and time (2020, 2021, etc.) you can use a nested loop to consider all the possible combinations i.e. the **cartesian product**
- Watch out for indentation!

````python
# Data lists for spatial directions and years
space = ['North', 'South', 'East', 'West']
years = [2020, 2021, 2022]
print(f"We have two lists: space {space} and years {years}")
print("We will use nested for loops to generate all possible combinations of space and time")
# Nested For Loops to generate combinations
for location in space: 
    for year in years:  #Single INDENTED! | is inside 1 for
        # Generate and print each space-time combination | DOUBLE INDENTED- is inside 2 fors
        print(f"Location: {location}, Year: {year}")
````

### The `range` Function

- Generates a sequence of numbers
    - You can acess its elements using indexing, slicing and striding
    - You can convert it to Lists, tuples, etc.
    - Commonly used to set the values of a `for` loop
- Syntax: `range(start, stop, step)`
    - `start`: Starting value (default is 0)
    - `stop`: Ending value (exclusive)
    - `step`: Increment (default is 1)
    - If called with just one par it is assumed to be the `stop`
        - `range(5)` is the same as `range(0, 5, 1)`
    - If called with two pars they are assumed to be `start` and `stop`
        - `range(2, 4)` is the same as `range(2, 4, 1)`

### Iterating Through Lists with `for` Loop Using Index and Element

- Using `range(len(iterable))` to create the loop sequence you can access both indexes and elements
    - `len(iterable)`: Returns the length of the iterable
    - `range()`: Generates indices from 0 to `len(iterable)-1`
    - You can use the indices to access the elements of the iterable
- Useful for tasks that require element manipulation based on index
    - Modify elements based on their index
    - Use the index along with the element



````python
# Use range and len to get both index and element
fruits = ['apple', 'banana', 'cherry']
print(f"Original List: {fruits}\n")  # print original list

# For Loop using range and len to iterate over the list
print(f"Using range and len to iterate over the list")
for index in range(len(fruits)):
    element = fruits[index]  # Access element using index
    print(f"Index: {index}, Element: {element}")  # print index and element
````

### Iterating Through Lists with `for` Loop Using Enumerate

- `enumerate()` is a more efficient and more readable alternative that allows to access both indexes and elements of a sequence
- Syntaxis  `enumerate(iterable, start=0)`
    - `iterable`: The sequence to iterate through
    - `start`: The starting index (default is 0)

````python
print(f"\nUsing enumerate to iterate over the list getting index and element")
for index, element in enumerate(fruits): # Using enumerate to get both index and element
    print(f"Index: {index}, Element: {element}")  # print index and element
````

### Outputs of a For Loop - Temporary or Lasting

- The output of a for loop can be temporary or lasting
    - Temporary: the output is not stored anywhere
        - Already covered this case in previous examples
    - Lasting: the output is stored in a variable

### Outputs of a For Loop - Lasting Output Common Patterns
- Very often you use one of these two patterns:
    1. **Var creation**: create a dedicated variable before the loop and then update it in the loop 
    1. **Var updating**: use a variable already existing and update it in the loop 
        - Possibly updating (overwriting) the iterable itself
        - You do not need to create new variables

### For Loops Outputting a Composite Data-Type - Var Creation
**Known Output Length - Pre-allocate**
- Example: Computing the cumulative sum (cumsum) of a list of numbers
    - Each entry is computed as the sum of all previous elements
- The steps of this example are:
    1. Initialize a list of length equal to the original list to store the output
    1. Loop through the list using `range` and `len` to retrieve the current element position and, at each iteration:
        1. Add the current element of the original list to the cumsum
        1. Store the cumsum in the output list at the position of the current element

### For Loops Outputting a Composite Data-Type - Var Creation


````python
#Computing cumulative sum with a pre-defined list
numbers = [1, 2, 3, 4, 5]
cumsum_fixed = [0] * len(numbers)#pre-allocate a list of zeros with the same length as numbers
cumulative_sum = 0 # initialize cumulative sum to zero
for index in range(len(numbers)):
    cumulative_sum += numbers[index] # add the current number to the cumulative sum
    cumsum_fixed[index] = cumulative_sum # store the cumulative sum
print(f"Cumulative Sum with Pre-defined Length: {cumsum_fixed}") # print the cumulative sum list
````

### For Loops Outputting a Composite Data-Type - Var Creation
**Unknown Output Length - Use extend/append**
- Example: Create a list with the elements of multiple lists characterized by random length
- The steps of this example are:
    1. Initialize an empty list to store the output
    1. Perform multiple iterations and, at each iteration:
        1. Randomly draw a number in [1, 3]
        1. Create a list_loop with length equal to the drawn number
        1. Extend the output list with the loop list

### For Loops Outputting a Composite Data-Type - Var Creation

````python
import random # import the random package
randomLen_outputList = []  # create an empty list to store results
for iter_loop in range(1, 6): # does 5 loops in [1, 5]
    # takes a random draw of a number in [1, 3] with equal probability
    randomLen_loopDraw = random.randint(1, 3) 
    # creates a list with length equal to randomLen_loop whose values are iter_loop
    randomLen_loopList = [iter_loop]*randomLen_loopDraw 
    # extend the output list with the elements of the loop list
    randomLen_outputList.extend(randomLen_loopList)
    # print the length at each loop
    print(f"The number of elements of iteration {iter_loop} is: {randomLen_loopDraw}")
print(f"The whole 'extended' list is {randomLen_outputList}")#Print the whole output
````

### Understanding Break and Continue in Nested Loops - Continue in Inner Loop

`continue` in **inner** loop skips current iteration of **inner** loop and continues with next iteration of **inner** loop

````python
print("Example 1: Using 'continue' in inner loop")
for i in range(3):
    for j in range(3):
        # if j=1 the code goes to next inner iteration (and skips printing)
        if j == 1: 
            continue 
        print(f"i: {i}, j: {j}")
````

### Understanding Break and Continue in Nested Loops - Break in Inner Loop
`break` in **inner** loop exits the innermost loop and starts the next iteration of the **outer** loop

````python
print("Example 2: Using 'break' in inner loop")
for i in range(3):
    for j in range(3):
        if j == 1: 
            break # if j=1 the code breaks the inner loop and goes to next outer iteration
        print(f"i: {i}, j: {j}")
````

### Understanding Break and Continue in Nested Loops - Continue in Outer Loop
`continue` in **outer** loop skips the current iteration of **outer** loop and continues with next iteration of **outer** loop

````python
# Using 'continue' in outer loop
print("Example 3: Using 'continue' in outer loop")
for i in range(3):
    if i == 1:
        continue # if i=1 the code goes to next outer iteration
    for j in range(3):
        print(f"i: {i}, j: {j}")
````

````python
# Using 'break' in outer loop
print("Example 4: Using 'break' in outer loop")
for i in range(3):
    if i == 1:
        break # if i=1 the code breaks the outer loop
    for j in range(3):
        print(f"i: {i}, j: {j}")
````

### While Loops - **Example: Savings Growth** 


````python
initial_balance = 1000  # Initial savings balance in USD
target_balance = 2000  # Target savings balance in USD
annual_interest_rate = 15  # Annual interest rate in percentage
years = 0  # Counter for years
print(f"Calculating time to reach a savings target of ${target_balance} with an initial balance of ${initial_balance} and an annual interest rate of {annual_interest_rate}%\n")
while initial_balance < target_balance: # While Loop to calculate the time to reach the target balance
    annual_interest = initial_balance * (annual_interest_rate / 100) # Calculate interest
    initial_balance += annual_interest # Update the current balance
    years += 1                         # Increment the year counter
    print(f"After {years} years, the balance is: ${initial_balance:.2f}")
# Print the total number of years needed to reach the target balance
print(f"It will take {years} years to reach a balance of ${target_balance}.")
````

## Dictionaries

### Dictionaries

- Dictionaries do not use indexes, like lists. They use so-called **keys**, mapped to **values**. 
- They are similar to a bilingual dictionary, e.g., 
    - From the English key 'one' to the German value 'ein',
    - From the English key 'two' to the German value 'zwei' etc.

### Dictionary Syntax rules

The syntax of a dictionary is:

    { ..., key: value, ....  }

- A `key` and its `value` are separated by a colon. They form an **item**.
- Each item `key: value` is separated commas

````python
# Create a dictionary
my_dict = {'one': 'ein', 'two': 'zwei', 'three': 'drei'}
# Display the dictionary
print(f"my_dict = {{'one': 'ein', 'two': 'zwei', 'three': 'drei'}} results in: \n{my_dict}")
````

### Rules for Keys and Values 

- Keys
    - Each `key` is **unique** so as to identify its own `value`. 
    - A key is **immutable** so: 
        - It can be an atomic type (e.g., a Boolean, an integer) or a string 
        - It cannot be a list or a dictionary (their length is not fixed, so mutable)
- Values
    - A `value` can be any type, including a list or a dictionary

### Dictionary Key Restrictions

````python
my_dict = { 
        'id' : 3232322,
        'school': 'Galilei', 
        'courses': [ 'English', 'Maths' ], 
        'students': { 'Sarah': 21, 'Bob': 19 } 
        }
print(my_dict)
````

````python
# keys must be immutable: lists are NOT accepted as keys
my_dict = {[ 'courses' , 'English', 'Maths' ]: { 'Sarah': 21, 'Bob': 19 } }
````

### Dictionary Key Restrictions

````python
# keys are immutable: dictionaries are NOT accepted as keys
my_dict = { 
        { 'courses' : [ 'English', 'Maths'] }: { 'Sarah': 21, 'Bob': 19 } 
        }

print(my_dict)
````
