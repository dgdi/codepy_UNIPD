### Data Types Characteristics
- **Scalar** vs **Non-scalar**
  - **Scalar**: Represents a single value
    - Simple so faster to process and manipulate
  - **Non-scalar**: Represents a collection or composite of multiple values
    - Can represent more complex data structures


### Data Types Characteristics
- **Mutable** vs **Immutable**
  - **Mutable**: The value can be changed after creation
    - Provides flexibility in data manipulation, allowing for modifications, additions, and deletions
  - **Immutable**: Once created, the value cannot be changed
    - Ensures data integrity, especially when passed between functions or threads, preventing unintentional modifications

### Data Types Characteristics
- **Ordered** vs **Unordered**
  - **Ordered**: Maintains a specific sequence or arrangement of elements
    - Predictable element arrangement allows for indexed access and sequence-based operations
  - **Unordered**: Does not maintain any specific sequence of elements
    - Typically faster for certain operations like membership tests and insertions since it doesn't need to maintain order

### Arithmetic Operations

The `+` operator between two numeric types computes their addition.

The `-` operator between two numeric types computes their subtraction.

The `*` operator between two numeric types computes their multiplication.

The `/` operator between two numeric types computes their division, returning their quotient as float number.

The `//` operator between two numeric types computes their integer division, returning their quotient as integer number.

The `%` operator between two numeric types computes their module, returning their remainder as integer number.

The `**` operator between two numeric types computes the exponentiation of the left-side by the power of the right-side number. 

````python
a = 3
b = 2

# Display 'a' and 'b'
print(f"The number a is {a}, the number b is {b}")

# Display the sum of 'a' and 'b'
print(f"a + b = {a + b}")

# Display the difference between 'a' and 'b'
print(f"a - b = {a - b}")

# Display the product of 'a' and 'b'
print(f"a * b = {a * b}")

# Display the division of 'a' by 'b'
print(f"a / b = {a / b}")
````

### Floating-Point Precision Issues

- Integers represent exactly whole numbers (much less demanding for the computer)
- Floating point numbers that are (very precise) **approximations** of a number
    - Representing number using powers of 2
    - Since doubles are approximations, comparisons of equality sometimes return unexpected results

````python
x_precision = 0.1
x_precision += 0.05
print(x_precision)
print(x_precision == 0.15)
````

The number `0.1` and `0.05` are stored with approximations, and their sum is not exactly `0.15`.




### Check True Value

- Already from printing `x_precision` you could see that something was off
    - To further inspect the true value, you can use the `format` function (more on that later)

````python
print(format(x_precision, ".50f"))
print(format(.15, ".50f"))
print(format(.1+.05, ".50f"))
````


This is why `==` returns `False`

 - Two numbers equal if the difference between the two is small enough

the `math` package provides an -equality check with tolerance- function already implemented

### Accessing Elements in Sequences (Characters in Strings) 

* Sequences: Lists, tuples, and **strings** are ordered
    - Each element in these data types is indexed by an integer 
    - You can access each element using the indexing operator `[]`
    
You can:
* **Index** `sequence[index]` to extract individual elements
* **Slice** `sequence[start:end]` to get sub-sequence starting from index `start` and ending at index `end-1`
* **Striding** `sequence[start:end:step]` to get sub-sequence starting from index `start`, in steps of `step`, ending at index `end-1`
  

 ### Indexing

Indexing is perfomed via square brackets and an integer
 - The index represents the position of the element to retrieve
  - **Indexing starts at 0**
  - Negative indices start from the end of the string    

#### Non-negative indexes

* Non-negative indexes are 0 or positive integers

* 0 indicates the first element

* An increase in the index moves one position from left to right

\begin{array}{|r||r|r|r|r|r|r|r|r|} 
  \hline
    \text{Non-negative index:} &  \phantom{-}0   &     \phantom{-}1    &    \phantom{-}2     &   \phantom{-}3     &      \phantom{-}4    &     \phantom{-}5     &     \phantom{-}6    &  \phantom{-}7       \\  \hline

````python
slides_string = "Hermione"
print(slides_string[0])
print(slides_string[1])
print(slides_string[5])
````

### Length of a String - The number of characters
You can count the number of characters in a string using the function `len`
- Unlike indexing that starts at 0, counting starts at 1
  - If you want to retrieve the last character in a string you need to index at `len(str)-1`
  - If you try to retrieve at an index that is not there, an exception will be raised

````python
print(len(slides_string))
print(slides_string[(len(slides_string)-1)])
print(slides_string[len(slides_string)])
````

### Slicing 

- Slicing can be performed on a variety of objects (strings, lists, NumPy arrays)
    - It is a way to extract a subset of elements from an object
- Slice notation uses `start:end` format within square brackets `[]`
  - Start index is included, stop index is excluded.
  - Result length: `stop - start`.
- If you omit one of the parameters uses defaults for `start` and `stop`:
    - `start`  defaults to `0`
    - `stop` defaults to `len(string)-1`

````python
print(slides_string[1:6])
print(slides_string[:6])
print(slides_string[6:])
````

### Striding 

Striding is performed via:
- Square brackets and double colons separating three fields
    - A field can be filled with either an integer or nothing
    - The left-most integer represents the start index (included)
    - the central integer reprents the end index (excluded)
    - the right-most integer represents the stride
        - The number of chars to move forward at each step
        - A negative number move backwards, eg., -1 reverses the order of the elements


$\begin{array}{|r||r|r|r|r|r|r|r|r|}  
  \hline
  \text{Element:} &  \text{H}  & \text{e} & \text{r} & \text{m}  & \text{i} & \text{o} & \text{n} & \text{e} \\ \hline
\end{array}
$

Then `word[::-1]` is `word` read backwards:

$
\begin{array}{|r||r|r|r|r|r|r|r|r|}  
  \hline
  \text{Element:} &  \text{e}  & \text{n} & \text{o} & \text{i}  & \text{m} & \text{r} & \text{e} & \text{H} \\ \hline
\end{array}
$

````python
print(slides_string[::2])
print(slides_string[2::2])
print(slides_string[::-1])
````

### Immutability of Python Strings
- If you need to modify components of a string you can't do it "in place" (in the same variable)
   - You need to modify the original object and then re-assign it (possibly to a variable with the same name, overwriting the original)
   - To use the original string as a starting point, you can use the `replace` method
       - Remember that a method is a function that is attached to an object (usually done for operations that are common for that object type)
       - In this case we want to apply the `replace` method to a string so the syntax will be `stringname.replace()`

````python
###just modify the string, without changing the original string
a = "this is a string"
print(a.replace("string", "longer string"))
print(a) ###Original string remains unchanged after modification.
###modify the original string and save it to a new variable
b = a.replace("string", "longer string") 
print(b)
###modify the original string and re-assign it to the same variable (overwrite/update the original string)
a = a.replace("string", "longer string") 
print(a)
````

### Formatting Decimal Digits with f-Strings in Python

- When you print a float, Python displays a certain number of decimal digits by default
- Using f-strings, you can control the number of decimal places by appending a format specification.
- Syntax: `{expression:.--num_of_digits--f}`
  - `expression`: The value you want to format.
  - `num_of_digits`: Number of decimal digits to display.
  


````python
pi = 3.14159265359
formatted_pi = f"Value of Pi rounded to 2 decimal places: {pi:.2f}"
formatted_pi_4_digits = f"Value of Pi rounded to 4 decimal places: {pi:.4f}"
print(formatted_pi)
print(formatted_pi_4_digits)
````

### Other Main Formatting Options in f-Strings

````python
name = "Alice"
f"Name: {name:s}"
````

### Scientific Notation
- `{variable:e}`: Formats the variable in scientific notation.

````python
number = 1000000
f"Scientific notation: {number:e}"
````

### Percentage Formatting
- `{variable:%}`: Formats the variable as a percentage.

````python
percent = 0.25
f"Percentage: {percent:%}"
````

### Combining Format Codes
- You can combine multiple format codes for a single variable.


````python
number = 12.3456
f"Combined: {number:.2%}"
````

- Clearly you can use a different formatting for different variables

````python
large_number = 123456789
completion_rate = 0.987
f"Large Number in Scientific Notation: {large_number:e}, Completion Rate: {completion_rate:.2%}"
````

### Encoding - Unicode and ASCII

- As we already mentioned, text is repesented on a computer using encodings
    - ASCII is the most basic (and old) encoding, using 7 bits to represent 128 characters
       - This implies that different languages assign different characters to the same bit patterns
       - Need to use the right encoding to display text correctly (italians use latin1, greeks use greek, etc.)
    - Unicode is a superset of ASCII and supports over a million characters
        - Standard for representing characters in Python (str)
        - Represents text not in terms of bites but with code points (integers)
        - No longer need to worry about the underlying encoding
- Sometimes it happens that when you read a file, the encoding is not correct
    - You will see strange characters instead of the expected ones
    - You can specify the encoding when opening the file (e.g., `open('file.txt', encoding='utf-8')`)

Truth Table of Negation:

|   `a`    | `not a` |
| -------  |  ------ |

| `False` | `False` | `False`     |

````python
#Second Table
print(f"True and True -> {True and True}")
print(f"True and False -> {True and False}")
print(f"False and True -> {False and True}")
print(f"False and False -> {False and False}")
````

For Disjunction:

### The "bit-wise" and "element-wise" Logical Operators

 Similar **but different** logical operators are:
  - `~`, `&`, and `|` that correspond to `not`, `and`, `or` but operate on each element of an object
  - If used on Python's numeric formats they act on each 0 and 1 storing its content (bit-wise)
      - As we said already, all the data in a computer is stored as 0s and 1s on a computer
      - It is sometimes useful to consider each 0/1 as True/False and operate on each of them using logical operators

### The Many Ways of Creating a Boolean - Comparison Operators

 Booleans can arise from the use of comparison operators  

Let `@` be any of the below comparison operators. Then, if the comparison has a sense (there is an ordering) the expression `a @ b` is Boolean

- In other words, the result `a @ b` is either `True` or `False`
- Ordering is cardinality for numbers (which one is bigger), alphabetical for strings
    - For other types it might be not defined, or defined by the programmer arbitrarily
    

| Operator | Example  |             Meaning          |	Result                                                              |
| -------- | -------- | ---------------------------- | -------------------------------------------------------------------- |   
| == 	   | `a` == `b`   | Equal to 	                 |  True if the value of `a` is equal to the value of `b` False otherwise   |
| != 	   | `a` != `b`   |	Not equal to 	             |	True if `a` is not equal to `b` False otherwise                         |
| < 	   | `a` < `b`    |	Less than 	                 |	True if `a` is less than `b` False otherwise                            |
| <= 	   | `a` <= `b`   |	Less than or equal to        |	True if `a` is less than or equal to `b` False otherwise                |
| > 	   | `a` > `b`    |	Greater than 	             |	True if `a` is greater than `b` False otherwise                         |
| >= 	   | `a` >= `b`   |	Greater than or equal to     |	True if `a` is greater than or equal to `b` False otherwise             |

````python
print(f"5==3 -> {5==3}")
print(f"5!=3 -> {5!=3}")
print(f"7<7 -> {7<7}")
print(f"5>3 -> {5>3}")
print(f"5>=5 -> {5>=5}")
````

````python
string_1 = "Hello"
string_2 = "Hi"
#comparison in strings is defined lexicographically (i.e. by the alphabetical order of subsequent chars)
print(f"{string_1} == {string_2} -> {string_1==string_2}")
print(f"{string_1} >= {string_2} -> {string_1>=string_2}")
print(f"{string_1} < {string_2} -> {string_1<string_2}")
````

### Formatting dates - Exporting dates to strings
- You can ask for the date-time information in the format you want
- The keys for the most common components are: 
     - `%Y` for the year
     - `%m` for the month
     - `%d` for the day
     - `%H` for the hour
     - `%M` for the minute
     - `%S` for the second 
- you can find all the components  [here](https://strftime.org/)


````python
print(f"dt.strftime('%Y-%m-%d %H:%M') -> {dt.strftime('%Y-%m-%d %H:%M')}")
print(f"dt.strftime('%Y/%m/%d %H.%M.%S')  -> {dt.strftime('%Y/%m/%d %H.%M.%S')}")
````

### Parsing dates - Importing dates from strings
- Often dates are stored in string format
    - We need to parse them into `datetime` objects
    - We can use the function `strptime` to parse 
    - Must provide the format of the date string
        - e.g., `%Y-%m-%d` for `2020-01-01` or `%d/%m/%Y` for `01/01/2020`

````python
date_strp_1 = datetime.strptime("20091031", "%Y%m%d")
date_strp_1.strftime("%Y-%m-%d %H:%M") 
````

````python
date_strp_2 = datetime.strptime("2009-11-30", "%Y-%m-%d")
date_strp_2.strftime("%Y-%m-%d %H:%M") 
````

````python
date_strp_3 = datetime.strptime("31122009", "%d%m%Y")
date_strp_3.strftime("%Y-%m-%d %H:%M") 
````

### Manipulating `datetime` Objects
- You can edit each component  of the detatime objects using the `replace`
    - Analogous to the `replace` method for strings



````python
print(f"dt.strftime('%Y-%m-%d %H:%M') -> {dt.strftime('%Y-%m-%d %H:%M')}")
dt_hour = dt.replace(minute=0, second=0)
print("dt_hour = dt.replace(minute=0, second=0)")
print(f"dt_hour.strftime('%Y-%m-%d %H:%M')  -> {dt_hour.strftime('%Y-%m-%d %H:%M') }")
````


- `datetime` objects are immutable; methods produce new objects.


### Working with Time Deltas
- Subtracting two `datetime` objects gives a `datetime.timedelta`.


````python
delta = dt - dt_hour 
delta
````

- Adding a `timedelta` to a `datetime` gives a new shifted datetime.

````python
print(f"dt_hour.strftime('%Y-%m-%d %H:%M')  -> {dt_hour.strftime('%Y-%m-%d %H:%M') }")
dt_original = dt_hour +  delta  
dt_2deltas = dt_hour +  2 * delta  
print("dt_original = dt_hour +  delta") 
print("dt_2deltas = dt_hour +  2 * delta")  
print(f"dt_original('%Y-%m-%d %H:%M')  -> {dt_original.strftime('%Y-%m-%d %H:%M') }")
print(f"dt_2deltas.strftime('%Y-%m-%d %H:%M')  -> {dt_2deltas.strftime('%Y-%m-%d %H:%M') }")


````

### Types and  Casting in Python

- The type of an object can be checked with the `type()` function

- It is also possible to convert values from one type to another with the following functions

| Function  |  	Description                                                         | 
| --------- | -------------                                                         | 
| bool() 	| Converts an argument to a Boolean value                               | 
| float() 	| Returns a floating-point object constructed from a number or string   | 
| int() 	| Returns an integer object constructed from a number or string         | 
| str() 	| Returns a string version of an object                                 | 

- Not all the conversions are possible and some may lead to loss of information (**always check the output!**)
    - `int` of a `float` will truncate the decimal part
    - `bool(0)` is False, but `bool` of any other number is True
    - String conversion usually does what you expect, but not always (e.g., string of a list is not the list itself)
    - `int` and `float` on a string 
        - Will work if the content can be parsed as number
        - Otherwise they raise an error

````python
float_var = 5.8
char_var = "Ciao"
charInt_var = "5"
print(f"float_var = {float_var}")
print(f"type of float_var is {type(float_var)}")
print(f"int(float_var) -> {int(float_var)}")
print(f"bool(0) -> {bool(0)}") 
print(f"bool(float_var) -> {bool(float_var)}") 
print(f"int(charInt_var) -> {int(charInt_var)}") 
print(f"int(char_var) -> {int(char_var)}") 
````
