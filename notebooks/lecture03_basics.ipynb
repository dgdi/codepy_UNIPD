### Functions

### Assignment

When programming you will need to store information inside **variables** (Tom's boxes of lecture 1, the memory locations)

  - Creates a binding between a name and an object

### Aliasing

- If you assign a new name to an existing variable, it does not create a clone of the data
  - It binds a new name to the same object, that now has an **alias** 
  - If `a` and `b` are assigned to the same object, we say that the object is **aliased** (it has two different names)
  

    

### Aliasing


````python
a = [1, 2, 3]                     # The name a is assigned to the list object [1,2,3]
b = a                             # The name b is assigned to the same list object referenced by a
print(f"a is: {a} and b is: {b}") # this is an f-string, more on that when talking about strings
````

<img src="img/alias.png" alt="assignment_pict" width="300"/>


 ### Modifying Aliased Objects
 
 Modifications made to an object (using one of the aliases) are going to be reflected on all the aliases


````python
a = [1, 2, 3]                     # The name a is assigned to the list object [1,2,3]
b = a                             # The name b is assigned to the same list object referenced by a
print(f"a is: {a} and b is: {b}")
a.append(4)                       # This *modifies* the list object referenced by both a and b
print(f"a is: {a} and b is: {b}") 
````

 <img src="img/append.png" alt="assignment_pict" width="300"/>


### Assignment: Modifications vs Re-Assignment 

- There are crucial differences between **modifying** an existing object
    - Such as the list appending in the previous slide `a.append(4)`  
- And **re-assigning** a variable
    - Such as re-assigning `a`, i.e., `a = an_object`
 - **Modification** changes the underlying object
     - Affecting all the variables referring to that object
 - **Re-assignment** changes the variable-object binding of one of the variable names
     - This has no impact on other variables (b is unaffected)

### Assignment: Modifications vs Re-Assignment 
- In the example below, re-assignment removes the initial binding and creates a new one
- This does not affect `b`


````python
a = [1, 2, 3]                     # The name a is assigned to the list object [1,2,3]
b = a                             # The name b is assigned to the same list object referenced by a
print(f"a is: {a} and b is: {b}")
a = [1, 2, 3, 4]                  # This *reassignes* a to the new object [1, 2, 3, 4]
print(f"a is: {a} and b is: {b}") #b is not affected
    
````

### Cloning

- If you want to do an **independent clone** of an object, creating a binding (done with the assignment operator `=`) it's not enough

### Cloning

- The simplest method to clone a list in Python is to utilize the "bare" slice operator `[:]` (the `[]` operators can do various things, we will see more in later lectures)
- You can also use the `copy()` function (from the `copy()` module) safer but may not be available

````python
b = a[:]       # make a clone using "bare" slice
print(f"a is: {a} and b is: {b}")

a.append(5)                       # This *modifies* the list object referenced only by a
print(f"a is: {a} and b is: {b}") # b is unaffected
````

### Equality (of content) is Different than Identity
To clarify it migh be useful to introduce two operators

* `==` checks if the values of two operands are equal or not
    * Is the content of the box labeled `a` the same of the one in box `d`
* `is` checks if the two operands refer to the same object
    * Are the two labes `a` and `e` referring to exactly the same box (memory location)?

````python
a = [1, 2, 3]  
d = a
e = a[:]
print(f"Are a and d equal? {a == d}")
print(f"Are a and d the same object? {a is d}")
print(f"Are a and e equal? {a == e}")
print(f"Are a and e the same object? {a is e}")
````

### Names conflicts and Namespaces

### Data Modification, Assignment and Copying

- Most analyses start from one source dataset and yield multiple outputs (tables, plots, intermediate transforms) through stepwise changes.  
- **Never modify the original in place**. You may need the untouched version for a later analysis 
- **Never** use just `df_for_plot = df`. This only creates a reference and can cause silent changes 
- Always create explicit copies with clear names, e.g., `df_for_plot = df.copy()`


````python
a = [1, 2, 3]
print(f"a is: {a}")
c = {"year" : 1955, "month" : 11, "day" : 5} # a dictionary
a = c
print(f"now a is: {a}")
````

- A data analysis leverages lots of libraries and functions
    - Sometimes the same name might be used for different variables, functions, etc.
    - E.g., a variable ```variance``` defined in your script and a variable with the same name in a library
- How can python manage such name conflicts?


### Namespaces 
- Python implements a set of hierarchical **Namespace**
    - A set of mappings between names and objects
    - In each namespace all these names are unique and can be used without any conflict
    - Namespaces are implemented as dictionaries (names are keys, objects are values; more on dictionaries in next lectures)
- There are multiple namespaces in a Python program  
    - Local: Inside a function
    - Enclosing scope: Any enclosing functions
    - Global: Top-level of the module
    - Built-in: Provided by the builtins module
- You can view the namespace with the `locals()` and `globals()` functions




### Scope searching, or how to search across namespaces

- When you use a name, Python searches for it in the following order:
    - Local scope: The current function
    - Enclosing scope: Any enclosing functions
    - Global scope: The top-level of the module
    - Built-in scope: Provided by the builtins module
- Python runs the first object it finds
- Sometimes the coder messes up and the variables that is used by the code is not the intended one
    - A mistake that is very insidious and hard to identify
    - Very important to use descriptive and unique names for variables

### Scope creation

- What happens to the namespace when you import a module?
    - A new namespace is created for that module
    - You access the objects in that namespace using dot notation, like module_name.object_name (e.g., np.array $\rightarrow$ the function array in the numpy library that has been imported as np) 

- What happens when you call a function? 
    - A new local and **temporary** namespace is created for that function execution
    - This is where all the names local to the function will exist


### Practical implication of scoping - Use **distinctive and informative** names

- Good practice to use distinct, informative names
    - Crucial to avoid accidental confusion of vars names
    - It will help you and others understand your code when you read it again

    - `classroom_data_2` uses the **snake case** convention

# Filepath and a Folder Setup for Data Analysis

### How to Refer to a File in a Computer

- The hard drive is where all the computer files are stored
 - How can the computer find a specific file?
    - To find something usually we need a position or an address
    - The file path is the address of a file
- An address in a computer has a tree structure
    - The root is the hard drive (e.g., `C:` on Windows, `/` on Mac)
    - The branches are the folders (e.g., Desktop, Documents, etc.)
    - The leaves are the files (e.g., lecture03_basics.ipynb)
- The parent folder of a file/folder is the one that contains it
    - A child folder is contained by the parent folder

### Relative and Absolute Paths

- The working directory is the folder where the kernel is currently looking for files
    - e.g., `C:\Users\dgdi\projects_py\codepy\notebooks`
    - This is an absolute path: the full path to a file starting from the HD
- A path is instead called relative if it starts from the current working directory
    - If we are in the `codepy` folder, we can refer to the lecture03_basics.ipynb file as `notebooks\lecture03_basics.ipynb`

### Limits of Absolute Paths

- Absolute paths are fine if you work on just one computer
    - Other computers might have different folder structure (e.g., a different user name)
    - On other computers your code will not be able to find the files it needs to run
- This problem can be solved by using relative paths and setting at the beginning of the code the working directory
- Let's see how to setup a project folder that allows for a simple implementation of this principle

### Getting Info From The Error

When a Python script fails, usually it will raise an exception
   - The console will display the *traceback* with the information on the cause of the exception
     - Detailing the chain of function calls leading to the error
     - Somewhere in that chain there is the problem
   - Use `%xmode` (*exception mode*) to control the amount of information printed when the exception is raised

Before seeing a simple example, let's introduce the family of `%xmode` magic commands:

### IPython Magic Commands

- IPython entails *magic commands* that are prefixed by the `%` character
- Magic commands succinctly solve various common problems in data analysis, they come in two flavors: 
    - *line magics*, prefixed by a single `%`, that operate on a single line of input
    - *cell magics*, prefixed by a double `%%`, that operate on multiple lines of inputs
- There are many, we only review a couple to get an idea 

### Two functions
Here we define two functions: 
- `func1` that takes one argument `x`, creates the variable `a = x` and the variable `b = x - 1` and then applies func1
- `func2` takes two arguments `a` and `b` and computes `a/b`


````python
def func1(x):
    a = x
    b = x - 1
    return func2(a, b)

def func2(a, b):
    return a / b

````

### Controlling Exceptions: `%xmode`

````python
func1(1)
````

### Default Traceback 

- Calling `func2` results in an error, coming from its 7th line, coming from `func1`, during the division at line 2
- By default (`Context`), traceback shows each step leading to the error and the lines where the exception is raised
- `%xmode` takes one of three possibilities: `Plain`, `Context`, and `Verbose`.
   - The default is `Context`
   - `Plain` is more compact and gives less information:
   - The `Verbose` mode adds some extra information, including the arguments to any functions

````python
%xmode Plain
func1(1)
````

With `Plain` we no longer see the code where the exception is raised

````python
%xmode Verbose
func1(1)
````

Using the `Verbose` option we also have the information on the values of variables

### Error information

The extra information can help you understand why the exception is being raised

Which traceback option is best depends on the complexity of the code you are debugging
 - `Verbose` provides a lot of information but the traceback can get extremely long
 - Depending on the context, the brevity of `Plain` or `Context` may be easier to work with.

### Debugging: When Reading Tracebacks Is Not Enough

The IPython-enhanced version of this is `ipdb`, the IPython debugger.
   - This debugger lets you step through the code line by line 
   - Many ways to launch/usee the debuggers
       - The most convenient interface to debugging is the `%debug` magic command
       - Refer to the online documentation to learn more
   - If you run `%debug` after an exception is raised, it will  open an interactive debugging prompt at the point of the exception.
   - The `ipdb` prompt lets you explore the current state of the stack (the "pile" of commands that you have sent to be executed), explore the available variables, and even run Python commands!

### Debugging with AI

- What AI can do
    - Explain error messages and traceback in plain language
    - Suggest likely causes and concrete fixes (code patches, parameter changes)
    - Propose debugging steps (print checks, assert statements, logging)

- Best practices when asking an AI
    - Provide your code and the error output
    - Ask for multiple hypotheses and a prioritized list of fixes



### Debugging with AI

- Limitations & safety
    - Validate any AI-suggested change locally and with tests — do not apply blindly
    - Be careful sharing sensitive data (private data, login information) in prompts
    - Combine AI suggestions with human review and automated tests

- Workflow once you get an error
    1. copy the input and output (error message, traceback)
    2. Ask AI for explanations, hypotheses, and patches
    3. If you do not like the answer, generate a minimal reproducible example to put on SO
    4. Apply changes incrementally and run tests
    5. Review and document the fix
