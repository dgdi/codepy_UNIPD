## Part 2: Lists

### What is a List?

A **list** is a data structure that stores multiple values in a single object.

- Lists can contain mixed data types (numbers, text, booleans, etc.)
- In this lecture, we focus on lists of numbers

**Syntax:**

```python
[element_1, element_2, ..., element_n]
```

````python
# Create a list of numbers
lst = [1, 2, 3, 4, 5]
lst
````

### Indexing: Accessing List Elements

- Lists are **ordered collections** where each element has a position
- Use the `[]` operator to **index** (retrieve) elements by position
- **Position 0** is the first element, **position 1** is the second, and so on
- Positions are **zero-indexed** (first element is at index 0, not 1)

**Example:** A list of revenues for three days:

\begin{array}{|r|r|r|} 
\hline
\text{Position:} 
&  \phantom{-}0   
&  \phantom{-}1    
&  \phantom{-}2
\\ \hline
\text{Element:} 
& \text{10000}  
& \text{20000} 
& \text{30000}
\\ \hline
\end{array}

````python
# Define a list of revenues
revenues = [10000, 20000, 30000]

# Retrieve the element at index 1 (the second element)
second_revenue = revenues[1]
second_revenue
````

### Computing with List Elements

Once we extract an element from a list using indexing, we can use it in any calculation.

**Problem:** Suppose we have three revenues and a 22% tax rate. How do we compute the total tax?

**Approach:**
1. Extract each revenue using indexing
2. Multiply each revenue by the tax rate
3. Sum all the tax amounts

````python
# Example: compute total tax on three revenues
# tax rate
t = 0.22  
# revenues for three different products
tax_1 = revenues[0] * t
tax_2 = revenues[1] * t
tax_3 = revenues[2] * t
# total tax is the sum of the three taxes
total_tax = tax_1 + tax_2 + tax_3
total_tax
````

## Part 3: Iteration with For Loops

### What is a For Loop?

A **for loop** automates the execution of the same code block for each element in a sequence.

- Saves time and reduces errors
- Applies operations to all elements consistently
- Works with lists, ranges, and other sequences

### For Loop Syntax

```python
for variable in sequence:
    body statement ...
    body statement ...
    ...
```

**Key Elements:**

- `variable`: a name that takes each value from the sequence (e.g., `revenue`, `i`, `item`)
- `sequence`: any iterable (list, range, etc.)
- `body`: the code block that allows to manipulate each element 
    - **Indentation:** the loop body must be indented (Python uses indentation to define code blocks)
    - in the example below, the variable is displayed only, but it can be used in any operation


````python
# Example: print the first 5 revenues
for revenue_loop in revenues_long:
    print(revenue_loop)
````

### Use a for loop to compute total taxes

````python
# Compute total tax on all 100 revenues using a for loop
tax_rate = 0.2
total_tax_long = 0

for revenue in revenues_long:
    total_tax_long = total_tax_long + (revenue * tax_rate) # at each iteration, we add the tax on the current revenue to the total tax

total_tax_long
````

## Part 4: The range() Function

### Overview

`range()` generates a sequence of numbers that can be used in for loops.

**Syntax:**
```python
range(start, stop, step)
```

**Parameters:**

- `start`: Starting value (default: 0)
- `stop`: Ending value (exclusive, not included)
- `step`: Increment between values (default: 1)

**Note:** Use `list()` to convert the output of `range()`, otherwise it returns an iterable (we'll see them in the next lectures).

### Examples of range()

````python
list(range(1, 11, 1))
````

````python
list(range(0, 10000, 500))
````
