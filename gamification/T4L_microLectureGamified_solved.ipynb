### 🔄 What is a For Loop?

A **for loop** is like assigning a rookie officer to go through every file in the cabinet, one by one, and do the same thing to each.

```python
for item in sequence:
    do something with item ...
```

**Key elements:**
- `item`: a temporary name that takes each value from the sequence
- `sequence`: the list (or range) to loop over
- The **indented body**: the operation that runs once per item

```
  📁 Cabinet of 100 files
  │
  ├── File #1  →  Rookie reads it, does the task
  ├── File #2  →  Rookie reads it, does the task
  ├── File #3  →  Rookie reads it, does the task
  │    ...
  └── File #100 → Rookie reads it, does the task
```

**Example:** Print every stolen amount in the database

````python
# The rookie reads out every record
for amount in stolen_amounts:
    print(amount)
````

### 📊 The Accumulator: Building a Running Total

To **add up** all 100 stolen amounts, we use the **accumulator pattern**:

1. Start with a total of 0 (empty evidence bag 👜)
2. Go through each heist record
3. Add the amount to the running total

```
  total = 0             ← empty evidence bag
  ┌───────────┐
  │ Heist #1  │ ──→  total = 0 + 12000 = 12000
  │ Heist #2  │ ──→  total = 12000 + 18000 = 30000
  │ Heist #3  │ ──→  total = 30000 + 15000 = 45000
  │   ...     │      ...
  │ Heist #100│ ──→  total = ??? (you tell me!)
  └───────────┘
```

**Example:** Total reward budget across all 100 heists

````python
# Compute the total reward fund (22% of each heist) across all 100 cases
reward_rate = 0.22
total_reward_100 = 0

for amount in stolen_amounts:
    total_reward_100 = total_reward_100 + (amount * reward_rate)

total_reward_100
````
